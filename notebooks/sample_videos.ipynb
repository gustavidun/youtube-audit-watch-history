```python
import polars as pl
from pathlib import Path

ROOT = Path("../")
```

Load watch histories and filter out shorts.

```python
shorts = pl.scan_csv(ROOT / "data/watch_histories/is_short_full.csv")
wh_pipe = (
    pl.scan_csv(
        ROOT / "data/watch_histories/video_histories.csv",
        schema_overrides={
            "video_id": pl.String,
            "Participant ID": pl.String,
        },
        try_parse_dates=True
    )
    .join(shorts, on="video_id")
    .filter(~pl.col("is_short")
    )
    .rename({"Participant ID": "participant_id"})
    .select([
        "video_id",
        "participant_id",
        "time"
    ])
)
wh_pipe.head().collect()
```

```python
wh_pipe.sink_parquet(ROOT / "data/watch_histories/video_histories_filtered.parquet")
```
